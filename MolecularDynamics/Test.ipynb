```python
import torch
import pykeops
from pykeops.torch import LazyTensor
import sys

 
print(f"Python: {sys.executable}")
print(f"GPU: {torch.cuda.get_device_name(0)}")
 

try:
    print("\nStarte Rechnung...")
    
    device = 'cuda'
    N = 5000
    x = torch.randn(N, 3, device=device)
    y = torch.randn(N, 3, device=device)
 
    x_i = LazyTensor(x[:, None, :])
    y_j = LazyTensor(y[None, :, :])
    
    print("   Kompiliere Kernel (das dauert beim ersten Mal ein paar Sekunden)...")
    D_ij = ((x_i - y_j) ** 2).sum(-1)
    res = D_ij.sum(0)
    
    print("✅ ERFOLG! KeOps hat kompiliert und gerechnet.")
    print(f"   Ergebnis-Tensor liegt auf: {res.device}")
    print(f"   Erste 3 Werte: {res[:10].cpu().numpy()}")

except Exception as e:
    print(f"\n❌ FEHLER: {e}")
```
